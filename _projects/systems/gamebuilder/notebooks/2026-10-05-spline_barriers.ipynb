### What are spline barriers?
A spline barrier is a smooth boundary in a game that blocks player movement. It is shaped by control points, like placing pins along a flexible fence and letting the fence curve between them.

A point has an `x` and `y` position: `{ x: 0.25, y: 0.6 }`. In GameBuilder, these positions are normalized: `0` is the left or top edge, `1` is the right or bottom edge.

A spline barrier needs at least two points. More points let you shape more bends in the barrier.

### Why are spline barriers helpful? 
Instead of placing many separate straight walls, a spline barrier uses a short list of points to make one smooth boundary. The game engine uses that curve for collision, so the player cannot move through it.

### 1. Control points

A control point is one position that helps shape the curve. Points are written in order, from one end of the barrier to the other.

**Examples:** a curved riverbank, a winding cave wall, or a boundary around a garden

`const points = [{ x: 0.1, y: 0.3 }, { x: 0.5, y: 0.7 }, { x: 0.9, y: 0.3 }];`


### Data Types 

**A data type is what kind of value is stored. Here are the main types used to describe a spline barrier:** 

`let point = { x: 0.25, y: 0.7 };` Object → one control point with x and y coordinates

`let points = [{ x: 0.2, y: 0.1 }, { x: 0.5, y: 0.8 }];` Array → an ordered list of points

`let isVisible = true;` Boolean → whether the barrier is drawn; a hidden barrier still blocks movement

`let barrierName;` Undefined → declared but not filled yet

`let noBarrier = null;` Null → intentionally empty

### Reference Data Types (Bigger containers) 

**A barrier combines related information and multiple control points.**

Object → a bundle of related barrier information
Array → an ordered list of control points

**Objects**

Instead of keeping each piece separately: 
`let name = "Garden Wall";` `let visible = true;` `let points = [{ x: 0.2, y: 0.3 }, { x: 0.8, y: 0.4 }];` 

Group the settings into one object: 
`const barrier = { name: "Garden Wall", visible: true, points: [{ x: 0.2, y: 0.3 }, { x: 0.8, y: 0.4 }] };` 

**Arrays:**

Instead of storing every coordinate as a separate variable:
`let point1 = { x: 0.1, y: 0.3 };` `let point2 = { x: 0.5, y: 0.6 };` `let point3 = { x: 0.9, y: 0.3 };`

Put the points into one ordered array:
`let points = [{ x: 0.1, y: 0.3 }, { x: 0.5, y: 0.6 }, { x: 0.9, y: 0.3 }];`

## HTML DOM (Document Object Model)

The HTML DOM is the live page in the browser. JavaScript can change the DOM to show information about a barrier, such as its control points. In a game, the engine also draws the curve and checks whether the player collides with it.

### So how does HTML DOM connect to spline barriers?

JavaScript stores a barrier's control points in an array. The page can read that array and display the coordinates. When a button adds a point, JavaScript updates the array and the DOM displays the updated list.

```python
%%html

<!-- UI_RUNNER: DOM change -->

<!-- Add a control point and update the displayed list. -->

<button onclick="addPoint()">Add Barrier Point</button>
<div id="barrierPoints">0.20, 0.40 | 0.50, 0.70</div>

<script>
let barrierPoints = [
  { x: 0.2, y: 0.4 },
  { x: 0.5, y: 0.7 }
];

function addPoint() {
    const nextPoint = { x: 0.8, y: 0.4 };
    barrierPoints.push(nextPoint);
    document.getElementById("barrierPoints").textContent = barrierPoints
      .map(point => `${point.x.toFixed(2)}, ${point.y.toFixed(2)}`)
      .join(" | ");
}
</script>
```

1. The HTML button calls a function, and the `<div id="barrierPoints">` is where the point list appears.

2. The click calls `addPoint()`.

3. The function adds a point and refreshes the displayed list:

  - add the new control point → `barrierPoints.push(nextPoint)`
  - update the DOM with the coordinates → `document.getElementById("barrierPoints").textContent = ...`
